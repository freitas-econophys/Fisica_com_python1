```python
# Reproduzindo o exemplo com adaptações:

# Q6) Momentos de distribuições de variáveis discretas.
import numpy as np
from scipy.stats import binom, boltzmann, nbinom, planck, poisson, bernoulli
import matplotlib.pyplot as plt

```

```python
# Suporte da distribuição binomial
n, p = 5, 0.4
lb_binom, ub_binom = binom.support(n, p)

# Determinando as grandezas estatísticas:
var_binom, skew_binom, kurt_binom = binom.stats(n, p, moments='vsk') # v -> variância; s -> skewness; k -> kurtosis.
```

```python
# Suporte da distribuição de Boltzmann
lambda_boltzmann, N = 1.5, 20
lb_boltzmann, ub_boltzmann = boltzmann.support(lambda_boltzmann, N)

#Definindo quantidades estatíticas
var_boltzmann, skew_boltzmann, kurt_boltzmann = boltzmann.stats(lambda_boltzmann, N, moments='vsk')
```

```python
# Suporte da distribuição binomial negativa (variáveis aleatórias)
n_nbinom, p_nbinom = 10, 1
lb_nbinom, ub_nbinom = nbinom.support(n_nbinom, p_nbinom)

#Definindo quantidades estatíticas
var_nbinom, skew_nbinom, kurt_nbinom = nbinom.stats(n_nbinom, p_nbinom, moments='vsk')
```

```python
# Suporte da distribuição de Planck
lambda_planck = 0.5
lb_planck, ub_planck = planck.support(lambda_planck)

#Definindo quantidades estatíticas
var_planck, skew_planck, kurt_planck = planck.stats(lambda_planck, moments='vsk')
```

```python
# Suporte da distribuição de Poisson
mu = 0.5
lb_poisson, ub_poisson = poisson.support(mu)

# Definindo as grandezas estatísticas
var_poisson, skew_poisson, kurt_poisson = poisson.stats(mu, moments='vsk')
```

```python
# Suporte da distribuição de Bernoulli
p = 0.700
lb_bernoulli, ub_bernoulli = bernoulli.support(p)

# Definindo as grandezas estatísticas
var_bernoulli, skew_bernoulli, kurt_bernoulli = bernoulli.stats(p, moments='vsk')
```

```python
# Montando o grid (matriz) 2X3 para os painéis

fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(20, 15))


# Distribuição binomial
x = np.arange(
    binom.ppf(0.01, n, p),
    binom.ppf(0.99, n, p) + 1
)

ax[0, 0].stem(
    x, binom.pmf(x, n, p),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 0].set_title('Distribuição Binomial')


# Distribuição de Boltzmann
x = np.arange(
    boltzmann.ppf(0.01, lambda_boltzmann, N),
    boltzmann.ppf(0.99, lambda_boltzmann, N) + 1
)

ax[0, 1].stem(
    x, boltzmann.pmf(x, lambda_boltzmann, N),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 1].set_title('Distribuição de Boltzmann')


# Distribuição binomial negativa
x = np.arange(
    nbinom.ppf(0.01, n_nbinom, p_nbinom),
    nbinom.ppf(0.99, n_nbinom, p_nbinom) + 1
)

ax[0, 2].stem(
    x, nbinom.pmf(x, n_nbinom, p_nbinom),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 2].set_title('Distribuição Binomial Negativa')


# Distribuição de Planck
x = np.arange(
    planck.ppf(0.01, lambda_planck),
    planck.ppf(0.99, lambda_planck) + 1
)

ax[1, 0].stem(
    x, planck.pmf(x, lambda_planck),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 0].set_title('Distribuição de Planck')


# Distribuição de Poisson
x = np.arange(
    poisson.ppf(0.01, mu),
    poisson.ppf(0.99, mu) + 1
)

ax[1, 1].stem(
    x, poisson.pmf(x, mu),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 1].set_title('Distribuição de Poisson')


# Distribuição de Bernoulli
x = np.arange(
    bernoulli.ppf(0.01, p),
    bernoulli.ppf(0.99, p) + 1
)

ax[1, 2].stem(
    x, bernoulli.pmf(x, p),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 2].set_title('Distribuição de Bernoulli')


plt.tight_layout()
plt.show()
```

```python
print('Binomial:')
print('Variância =', var_binom)
print('Skewness =', skew_binom)
print('Kurtosis =', kurt_binom)

print('\nBoltzmann:')
print('Variância =', var_boltzmann)
print('Skewness =', skew_boltzmann)
print('Kurtosis =', kurt_boltzmann)

print('\nBinomial negativa:')
print('Variância =', var_nbinom)
print('Skewness =', skew_nbinom)
print('Kurtosis =', kurt_nbinom)

print('\nPlanck:')
print('Variância =', var_planck)
print('Skewness =', skew_planck)
print('Kurtosis =', kurt_planck)

print('\nPoisson:')
print('Variância =', var_poisson)
print('Skewness =', skew_poisson)
print('Kurtosis =', kurt_poisson)

print('\nBernoulli:')
print('Variância =', var_bernoulli)
print('Skewness =', skew_bernoulli)
print('Kurtosis =', kurt_bernoulli)
```

```python
#Esse comando irá trocar "nan" por "momento não definido" (MND)
def formatar_momento(valor):
    if np.isnan(valor):
        return 'MND'
    else:
        return f'{valor:.3f}'
```

```python
# Montando o grid (matriz) 2X3 para os painéis com os títulos

fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(15, 8))


# Distribuição binomial
x = np.arange(
    int(binom.ppf(0.01, n, p)),
    int(binom.ppf(0.99, n, p)) + 1
)

ax[0, 0].stem(
    x, binom.pmf(x, n, p),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 0].set_title(
    f'Binomial\nVariância = {formatar_momento(var_binom)} | '
    f'Skewness = {formatar_momento(skew_binom)} | '
    f'Kurtosis = {formatar_momento(kurt_binom)}'
)


# Distribuição de Boltzmann
x = np.arange(
    int(boltzmann.ppf(0.01, lambda_boltzmann, N)),
    int(boltzmann.ppf(0.99, lambda_boltzmann, N)) + 1
)

ax[0, 1].stem(
    x, boltzmann.pmf(x, lambda_boltzmann, N),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 1].set_title(
    f'Boltzmann\nVariância = {formatar_momento(var_boltzmann)} | '
    f'Skewness = {formatar_momento(skew_boltzmann)} | '
    f'Kurtosis = {formatar_momento(kurt_boltzmann)}'
)


# Distribuição binomial negativa
x = np.arange(
    int(nbinom.ppf(0.01, n_nbinom, p_nbinom)),
    int(nbinom.ppf(0.99, n_nbinom, p_nbinom)) + 1
)

ax[0, 2].stem(
    x, nbinom.pmf(x, n_nbinom, p_nbinom),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[0, 2].set_title(
    f'Binomial Negativa\nVariância = {formatar_momento(var_nbinom)} | '
    f'Skewness = {formatar_momento(skew_nbinom)} | '
    f'Kurtosis = {formatar_momento(kurt_nbinom)}'
)


# Distribuição de Planck
x = np.arange(
    int(planck.ppf(0.01, lambda_planck)),
    int(planck.ppf(0.99, lambda_planck)) + 1
)

ax[1, 0].stem(
    x, planck.pmf(x, lambda_planck),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 0].set_title(
    f'Planck\nVariância = {formatar_momento(var_planck)} | '
    f'Skewness = {formatar_momento(skew_planck)} | '
    f'Kurtosis = {formatar_momento(kurt_planck)}'
)


# Distribuição de Poisson
x = np.arange(
    int(poisson.ppf(0.01, mu)),
    int(poisson.ppf(0.99, mu)) + 1
)

ax[1, 1].stem(
    x, poisson.pmf(x, mu),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 1].set_title(
    f'Poisson\nVariância = {formatar_momento(var_poisson)} | '
    f'Skewness = {formatar_momento(skew_poisson)} | '
    f'Kurtosis = {formatar_momento(kurt_poisson)}'
)


# Distribuição de Bernoulli
x = np.arange(
    int(bernoulli.ppf(0.01, p)),
    int(bernoulli.ppf(0.99, p)) + 1
)

ax[1, 2].stem(
    x, bernoulli.pmf(x, p),
    linefmt='r-', markerfmt='ro', basefmt='k-'
)

ax[1, 2].set_title(
    f'Bernoulli\nVariância = {formatar_momento(var_bernoulli)} | '
    f'Skewness = {formatar_momento(skew_bernoulli)} | '
    f'Kurtosis = {formatar_momento(kurt_bernoulli)}'
)


plt.tight_layout()
plt.show()
```
