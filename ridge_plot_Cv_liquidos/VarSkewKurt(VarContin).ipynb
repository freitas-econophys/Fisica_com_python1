```python
# Reproduzindo o exemplo com adaptações:

# Q5) Momentos de distribuições de variáveis contínuas.
import numpy as np
from scipy.stats import norm, pareto, cauchy, beta, landau, lognorm
import matplotlib.pyplot as plt

```

```python
# Suporte da distribuição normal
lb_norm, ub_norm = norm.support()

# Determinando as grandezas estatísticas:
var_norm, skew_norm, kurt_norm = norm.stats(moments='vsk') # v -> variância; s -> skewness; k -> kurtosis.
```

```python
# Suporte da distribuição de Pareto
b = 2.50
lb_pareto, ub_pareto = pareto.support(b)

#Definindo quantidades estatíticas
var_pareto, skew_pareto, kurt_pareto = pareto.stats(b, moments='vsk')
```

```python
# Suporte da distribuição de Cauchy (variáveis aleatórias)
lb_cauchy, ub_cauchy = cauchy.support()

# Definindo as grandezas estatísticas
var_cauchy, skew_cauchy, kurt_cauchy = cauchy.stats(b, moments='vsk')
```

```python
# Suporte da distribuição beta
a, B = 2.31, 0.627
lb_beta, ub_beta = beta.support(a, B)

# Definindo as grandezas estatísticas
var_beta, skew_beta, kurt_beta = beta.stats(a, B, moments='vsk')
```

```python
# Suporte da distribuição de Landau

lb_landau, ub_landau = landau.support()

# Definindo as grandezas estatísticas
var_landau, skew_landau, kurt_landau = landau.stats(moments='vsk')
```

```python
# Suporte da distribuição beta
s = 0.700
lb_lognorm, ub_lognorm = lognorm.support(s)

# Definindo as grandezas estatísticas
var_lognorm, skew_lognorm, kurt_lognorm = lognorm.stats(s, moments='vsk')
```

```python
# Montando o grid (matriz) 2X3 para os painéis

fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(20, 15))

# Distribuição normal
x = np.linspace(norm.ppf(0.01),
                norm.ppf(0.99), 100)

ax[0, 0].plot(x, norm.pdf(x),
              'r-', lw=5, alpha=0.6, label='norm pdf')

# Distribuição de Pareto
x = np.linspace(pareto.ppf(0.01, b),
                pareto.ppf(0.99, b), 100)
ax[0, 1].plot(x, pareto.pdf(x, b),
       'r-', lw=5, alpha=0.6, label='pareto pdf')

# Distribuição de Cauchy
x = np.linspace(cauchy.ppf(0.01),
                cauchy.ppf(0.99), 100)
ax[0,2].plot(x, cauchy.pdf(x),
       'r-', lw=5, alpha=0.6, label='cauchy pdf')

# Distribuição beta
x = np.linspace(beta.ppf(0.01, a, B),
                beta.ppf(0.99, a, B), 100)
ax[1,0].plot(x, beta.pdf(x, a, B),
       'r-', lw=5, alpha=0.6, label='beta pdf')

# Distribuição de Landau
x = np.linspace(landau.ppf(0.01),
                landau.ppf(0.99), 100)
ax[1,1].plot(x, landau.pdf(x),
       'r-', lw=5, alpha=0.6, label='landau pdf')

# Distribuição lognormal
x = np.linspace(lognorm.ppf(0.01, s),
                lognorm.ppf(0.99, s), 100)
ax[1,2].plot(x, lognorm.pdf(x, s),
       'r-', lw=5, alpha=0.6, label='lognorm pdf')
```

```python
print('Normal:')
print('Variância =', var_norm)
print('Skewness =', skew_norm)
print('Kurtosis =', kurt_norm)

print('\nPareto:')
print('Variância =', var_pareto)
print('Skewness =', skew_pareto)
print('Kurtosis =', kurt_pareto)

print('\nCauchy:')
print('Variância =', var_cauchy)
print('Skewness =', skew_cauchy)
print('Kurtosis =', kurt_cauchy)

print('\nBeta:')
print('Variância =', var_beta)
print('Skewness =', skew_beta)
print('Kurtosis =', kurt_beta)

print('\nLandau:')
print('Variância =', var_landau)
print('Skewness =', skew_landau)
print('Kurtosis =', kurt_landau)

print('\nLognormal:')
print('Variância =', var_lognorm)
print('Skewness =', skew_lognorm)
print('Kurtosis =', kurt_lognorm)
```

```python
#Esse comando irá trocar "nan" por "momento não definido" (MND)
def formatar_momento(valor):
    if np.isnan(valor):
        return 'MND'
    else:
        return f'{valor:.3f}'
```

```python
# Montando o grid (matriz) 2X3 para os painéis com os títulos

fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(15, 8))

# Distribuição normal
x = np.linspace(norm.ppf(0.01),
                norm.ppf(0.99), 100)

ax[0, 0].plot(x, norm.pdf(x),
              'r-', lw=5, alpha=0.6, label='norm pdf')

ax[0, 0].set_title(
    f'Normal\nVariância = {formatar_momento(var_norm)} | '
    f'Skewness = {formatar_momento(skew_norm)} | '
    f'Kurtosis = {formatar_momento(kurt_norm)}'
)


# Distribuição de Pareto
x = np.linspace(pareto.ppf(0.01, b),
                pareto.ppf(0.99, b), 100)

ax[0, 1].plot(x, pareto.pdf(x, b),
              'r-', lw=5, alpha=0.6, label='pareto pdf')

ax[0, 1].set_title(
    f'Pareto\nVariância = {formatar_momento(var_pareto)} | '
    f'Skewness = {formatar_momento(skew_pareto)} | '
    f'Kurtosis = {formatar_momento(kurt_pareto)}'
)


# Distribuição de Cauchy
x = np.linspace(cauchy.ppf(0.01),
                cauchy.ppf(0.99), 100)

ax[0, 2].plot(x, cauchy.pdf(x),
              'r-', lw=5, alpha=0.6, label='cauchy pdf')

ax[0, 2].set_title(
    f'Cauchy\nVariância = {formatar_momento(var_cauchy)} | '
    f'Skewness = {formatar_momento(skew_cauchy)} | '
    f'Kurtosis = {formatar_momento(kurt_cauchy)}'
)


# Distribuição Beta
x = np.linspace(beta.ppf(0.01, a, B),
                beta.ppf(0.99, a, B), 100)

ax[1, 0].plot(x, beta.pdf(x, a, B),
              'r-', lw=5, alpha=0.6, label='beta pdf')

ax[1, 0].set_title(
    f'Beta\nVariância = {formatar_momento(var_beta)} | '
    f'Skewness = {formatar_momento(skew_beta)} | '
    f'Kurtosis = {formatar_momento(kurt_beta)}'
)


# Distribuição de Landau
x = np.linspace(landau.ppf(0.01),
                landau.ppf(0.99), 100)

ax[1, 1].plot(x, landau.pdf(x),
              'r-', lw=5, alpha=0.6, label='landau pdf')

ax[1, 1].set_title(
    f'Landau\nVariância = {formatar_momento(var_landau)} | '
    f'Skewness = {formatar_momento(skew_landau)} | '
    f'Kurtosis = {formatar_momento(kurt_landau)}'
)


# Distribuição lognormal
x = np.linspace(lognorm.ppf(0.01, s),
                lognorm.ppf(0.99, s), 100)

ax[1, 2].plot(x, lognorm.pdf(x, s),
              'r-', lw=5, alpha=0.6, label='lognorm pdf')

ax[1, 2].set_title(
    f'Lognormal\nVariância = {formatar_momento(var_lognorm)} | '
    f'Skewness = {formatar_momento(skew_lognorm)} | '
    f'Kurtosis = {formatar_momento(kurt_lognorm)}'
)


plt.tight_layout()
plt.show()

print('MND = Momento não definido')
```
