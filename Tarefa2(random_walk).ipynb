```python
#Tarefa 2) Random walk process

import random
import matplotlib.pyplot as plt

def randomwalk(x, time):

  Ts = [0]
  X = [0]

  for ts in range(1, time+1):

      Ts.append(ts)
      rand = random.choice(x)
      step = X[ts-1] + rand
      X.append(step)

  return Ts, X

#Exemplos dados no vídeo + 1 extra
Ts1, X1 = randomwalk([-1, 1], 50)
Ts2, X2 = randomwalk([-1, 1, -10], 50)
Ts3, X3 = randomwalk([-2, 2], 50)

#plotando os gráficos

plt.figure(figsize = (10,8))
plt.plot(Ts1, X1, label = '[-1, 1]')
plt.plot(Ts2, X2, label = '[-1, 1, -10]')
plt.plot(Ts3, X3, label = '[-2, 2]')
plt.xlabel('Time')
plt.ylabel('Position')
plt.title('Random Walks')
plt.legend()
plt.show
```
